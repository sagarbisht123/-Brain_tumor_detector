## 2. Formatting the data

The data is given in different *classes* but needs to be put in two different directories namely *train_dir* and *test_dir*.

```python
import os
import shutil
import random
import math

def creating_train_test_dir(source_dir, train_dir, test_dir):
    # Create train and test directories
    os.makedirs(train_dir, exist_ok=True)
    os.makedirs(test_dir, exist_ok=True)

    # Loop through each class folder
    for class_name in os.listdir(source_dir):
        class_path = os.path.join(source_dir, class_name)

        if os.path.isdir(class_path):
            # Create the same class folder structure in the train and test directories
            os.makedirs(os.path.join(train_dir, class_name), exist_ok=True)
            os.makedirs(os.path.join(test_dir, class_name), exist_ok=True)

            # Get all the image files in the class folder
            images = [img for img in os.listdir(class_path) if os.path.isfile(os.path.join(class_path, img))]
            random.shuffle(images)  # Shuffle the images to randomize the selection

            # Calculate split index
            total_images = len(images)
            train_count = math.floor(total_images * 0.9)
            train_images = images[:train_count]
            test_images = images[train_count:]

            # Copy the images to the respective train and test folders
            for image in train_images:
                shutil.copy(os.path.join(class_path, image), os.path.join(train_dir, class_name, image))

            for image in test_images:
                shutil.copy(os.path.join(class_path, image), os.path.join(test_dir, class_name, image))

            print(f"Class '{class_name}': {len(train_images)} train, {len(test_images)} test images")

    print("Dataset split completed successfully (90% train / 10% test per class)")

# Example usage
creating_train_test_dir("G:\\My Drive\\01 Projects\\01. Brain tumor\\brain tumor\\whole data", "G:\\My Drive\\01 Projects\\01. Brain tumor\\brain tumor\\train_90p",
                         "G:\\My Drive\\01 Projects\\01. Brain tumor\\brain tumor\\test_10p")
```
